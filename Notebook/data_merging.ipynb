# Data Concat - bu dataframelarni ustma- ust yoki yonma yon qilib birlashtirish  

```python
# Vertical concat 

import pandas as pd 

df1 = pd.DataFrame({
    "name": ["Ali", "Gani"],
    "age": [25, 34]
})

df2 = pd.DataFrame({
    "name": ["Halima", "Salima"],
    "age": [15, 24]
})

df = pd.concat([df1, df2], axis=0)
print(df)
```

```python
# Horizontal concat 

import pandas as pd 

df1 = pd.DataFrame({
    "name": ["Ali", "Gani"],
    "age": [25, 34]
})

df2 = pd.DataFrame({
    "name": ["Halima", "Salima"],
    "age": [15, 24]
})

df = pd.concat([df1, df2], axis=1)
print(df)
```

# Data Merging - bu datalarni umumiy ustunlar orqali birlashtirish

```python
df3 = pd.DataFrame({
    "id": [1, 2],
    "name": ["Hamid", "Laylijon"]
})

df4 = pd.DataFrame({
    "id": [1, 2],
    "salary": [1500, 2555]
})


df = pd.merge(df3, df4, on="id")
print(df)
```

# data mergingda bir nechta usullar mavjud 

- inner - faqat mos kelganlarni birlashtirish 
- left - chap tamonni birlashtirish 
- right - ong tamonni birlashtirish
- outer - hammasini birlashtirish 

```python
pd.merge(df3, df4, on="id", how="inner")
pd.merge(df3, df4, on="id", how="left")
pd.merge(df3, df4, on="id", how="right")
pd.merge(df3, df4, on="id", how="outer")
```
