# Practise: Video Preprocessing


## 1. Video Loading


```python
import cv2

video_path = 'business.mp4'
cap = cv2.VideoCapture(video_path)

if cap.isOpened():
    print('Video loaded successfully')
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    print('Total frames:', total_frames)
    print('FPS:', fps)
else:
    print('Could not open video. Check the file path.')
```

## 2. Create Folder And Make Frames


```python
import os

frames_folder = 'frames'
os.makedirs(frames_folder, exist_ok=True)
print('Folder ready:', frames_folder)

frame_positions = [10, 20, 30]

for i, pos in enumerate(frame_positions, start=1):
    cap.set(cv2.CAP_PROP_POS_FRAMES, pos)
    success, frame = cap.read()
    if success:
        frame_name = f'frame_{i:03d}.jpg'
        frame_path = os.path.join(frames_folder, frame_name)
        cv2.imwrite(frame_path, frame)
        print('Saved:', frame_path)
    else:
        print('Could not read frame at position', pos)

cap.release()
```

## 3. Labeling


```python
labels = {
    "frame_001.jpg": "walking",
    "frame_002.jpg": "running",
    "frame_003.jpg": "jumping"
}

print(labels)
```

## 4. Sample Labeling


```python
import shutil

dataset_folder = 'dataset'
class_names = ['walking', 'running', 'jumping']

for class_name in class_names:
    class_path = os.path.join(dataset_folder, class_name)
    os.makedirs(class_path, exist_ok=True)
    print('Created:', class_path)

for frame_name, label in labels.items():
    source_path = os.path.join(frames_folder, frame_name)
    destination_path = os.path.join(dataset_folder, label, frame_name)

    if os.path.exists(source_path):
        shutil.copy(source_path, destination_path)
        print(f'Copied {frame_name} -> {label}')
    else:
        print(f'File not found: {source_path}')

for label in os.listdir(dataset_folder):
    label_path = os.path.join(dataset_folder, label)
    if os.path.isdir(label_path):
        for img in os.listdir(label_path):
            print(label, img)
```

## 5. Frame Normalization


```python
normalized_frames = []

for frame_name in labels.keys():
    frame_path = os.path.join(frames_folder, frame_name)
    frame = cv2.imread(frame_path)

    if frame is not None:
        frame = frame / 255.0
        normalized_frames.append(frame)
        print(f'{frame_name} normalized, shape: {frame.shape}')
    else:
        print(f'Frame not found: {frame_path}')
```

```python
normalized_folder = 'normalized_frames'
os.makedirs(normalized_folder, exist_ok=True)

for frame_name in labels.keys():
    frame_path = os.path.join(frames_folder, frame_name)
    frame = cv2.imread(frame_path)

    if frame is not None:
        frame = frame / 255.0
        save_frame = (frame * 255).astype('uint8')
        save_path = os.path.join(normalized_folder, frame_name)
        cv2.imwrite(save_path, save_frame)
        print(f'Saved normalized frame: {save_path}')
    else:
        print(f'Frame not found: {frame_path}')
```

## 6. Resizing Frames


```python
resized_frames = []

for frame_name in labels.keys():
    frame_path = os.path.join(frames_folder, frame_name)
    frame = cv2.imread(frame_path)

    if frame is not None:
        frame = cv2.resize(frame, (224, 224))
        resized_frames.append(frame)
        print(f'{frame_name} resized, new shape: {frame.shape}')
    else:
        print(f'Frame not found: {frame_path}')
```

```python
resized_folder = 'resized_frames'
os.makedirs(resized_folder, exist_ok=True)

for frame_name in labels.keys():
    frame_path = os.path.join(frames_folder, frame_name)
    frame = cv2.imread(frame_path)

    if frame is not None:
        frame = cv2.resize(frame, (224, 224))
        save_path = os.path.join(resized_folder, frame_name)
        cv2.imwrite(save_path, frame)
        print(f'Saved resized frame: {save_path}')
    else:
        print(f'Frame not found: {frame_path}')
```

## 7. Batching Frames


```python
frames = []

for frame_name in labels.keys():
    frame_path = os.path.join(frames_folder, frame_name)
    frame = cv2.imread(frame_path)
    if frame is not None:
        frame = cv2.resize(frame, (224, 224))
        frame = frame / 255.0
        frames.append(frame)

batch = frames[0:32]
print('Total prepared frames:', len(frames))
print('Batch size:', len(batch))
```
